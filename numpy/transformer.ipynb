```python
#Hyperparams - reduced size from original paper for feasible implementation on a laptop
d_model = 256 
d_model = 128
n_heads = 4
seq_len = 64
n_layers = 2
batch_size = 16
```

```python
# Load Pride and Prejudice from Project Gutenberg
import urllib.request
import re

url = "https://www.gutenberg.org/cache/epub/1342/pg1342.txt"
response = urllib.request.urlopen(url)
raw_text = response.read().decode('utf-8-sig')

# Find where the actual book starts (after Gutenberg header)
# Look for "It is a truth universally acknowledged" - the famous opening line
start = raw_text.find("It is a truth universally acknowledged")
if start == -1:
    start = 0

# Find where the book ends
end = raw_text.find("*** END OF THE PROJECT GUTENBERG EBOOK")
if end == -1:
    end = len(raw_text)

text = raw_text[start:end].strip()

# Remove illustration markers and extra whitespace
text = re.sub(r'\[Illustration[^\]]*\]', '', text)
text = re.sub(r'\s+', ' ', text)
text = text.lower()

print(f"Loaded Pride and Prejudice: {len(text):,} characters")
print(f"\nFirst 300 chars:\n{text[:300]}")
```

```python
vocab = set(text.lower())
vocab_size = len(vocab)
print(vocab_size)
```

```python
embeddings = np.random.randn(vocab_size, d_model) * 0.01
char_to_idx = {ch: i for i, ch in enumerate(vocab)}
idx_to_char = {i: ch for i, ch in enumerate(vocab)}
```

```python
def encode(text):
    """Turns all elements in our text into our tokens"""
    return [char_to_idx[char] for char in text if char in vocab]
```

```python
def decode(encoded_text):
    """Returns tokenized text to natural language form"""
    return [idx_to_char[idx] for idx in encoded_text]
```
